```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
```

```python
spark = SparkSession.builder.appName('DataFrameOps').getOrCreate()
```

```python
data = [
(1,"Sawe",30),
(2,"Kejelcha",41),
(3,"Kiplimo",68)
]
```

```python
schema = StructType([
    StructField("id",IntegerType(),False),
    StructField("name",StringType(),False),
    StructField("time",IntegerType(),False)
])
```

```python
df = spark.createDataFrame(data, schema)
```

```python
df.show()
```

```python
# Select and Filtering Data Use Cases
```

```python
df.filter(df.time<40).show()
```

```python
df.filter(df.name == "Kejelcha").show()
```

```python
df.orderBy(df.time.desc()).show()
```

```python
df.groupBy("name").count().show()
```

```python
data2 = [
(1,"Kenya"),
(2,"Ethiopia"),
(3,"Uganda")
]
```

```python
schema = StructType([
    StructField("id",IntegerType(),True),
    StructField("country",StringType(),True),
])
```
