```python
import os
from pyspark.sql import SparkSession

# Initialize your Spark Session
spark = SparkSession.builder.master("local[*]").getOrCreate()

# Base path containing your unzipped files
base_path = "/content/brazilian-ecommerce/"

# List of all the CSV files in your folder
csv_files = [
    "olist_customers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_products_dataset.csv",
    "product_category_name_translation.csv",
    "olist_sellers_dataset.csv"
]

# Loop through each file to read, print schema, and show preview
for file_name in csv_files:
    full_path = os.path.join(base_path, file_name)

    # Visual separator for clean output layout
    print("=" * 60)
    print(f"PROCESSING FILE: {file_name}")
    print("=" * 60)

    # 1. Read the CSV file
    df = spark.read.csv(full_path, header=True, inferSchema=True)

    # 2. Print the Schema structure
    print(f"\n--- Schema for {file_name} ---")
    df.printSchema()

    # 3. Show the first 5 records
    print(f"\n--- First 5 Rows of {file_name} ---")
    df.show(5, truncate=False)
    print("\n\n")
```
